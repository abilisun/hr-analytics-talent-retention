```python
import os
import pandas as pd
import numpy as np
import kagglehub
```

```python
path = kagglehub.dataset_download("rhuebner/human-resources-data-set")
csv_path = os.path.join(path, "HRDataset_v14.csv")
```

```python
df_raw = pd.read_csv(csv_path)
```

```python
display(df_raw.head())
```

```python
missing_vals = df_raw.isnull().sum()
print(missing_vals[missing_vals > 0])
```

```python
df_clean = df_raw.copy()
```

```python
df_clean.columns = df_clean.columns.str.strip().str.lower().str.replace(' ', '_')
```

```python
date_cols = ['dateofhire', 'dateoftermination', 'dob', 'lastperformancereview_date']
for col in date_cols:
    if col in df_clean.columns:
        df_clean[col] = pd.to_datetime(df_clean[col], errors='coerce')
```

```python
str_cols = df_clean.select_dtypes(include=['object']).columns
for col in str_cols:
    df_clean[col] = df_clean[col].str.strip()
```

```python
df_clean['employment_status_clean'] = np.where(df_clean['dateoftermination'].notnull(), 'Terminated', 'Active')
```

```python
output_filename = 'cleaned_hr_dataset.csv'
df_clean.to_csv(output_filename, index=False)
```
